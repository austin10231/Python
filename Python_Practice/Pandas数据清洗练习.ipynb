```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Helen', 'Ian', 'Julia'],
    'Age': [25, 30, 35, 40, 28, 33, 27, 45, 31, 38],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'Marketing', 'IT', 'Marketing', 'Finance'],
    'Salary': [5000, 7000, 8000, 9000, 5200, 7800, 6100, 9500, 6200, 8800],
    'Experience': [2, 5, 8, 10, 3, 7, 4, 12, 6, 9],
    'Gender': ['F', 'M', 'M', 'M', 'F', 'M', 'F', 'F', 'M', 'F']
}

df = pd.DataFrame(data)
df
```

```python
df.head(5)
```

```python
df.loc[(df['Department'] == 'IT') & (df['Salary'] > 8000)]
```

```python
df.groupby('Department').size()
```

```python
df.groupby('Department').agg({'Salary': 'mean', 'Age': 'mean'})
```

```python
def category(row):
    if row['Experience'] >= 8:
        return 'Senior'
    elif row['Experience'] >= 4:
        return 'Mid'
    else:
        return 'Junior'

df['Seniority'] = df.apply(category, axis=1)
```

```python
df.head()
```

```python
df.nlargest(3, 'Salary')[['Name', 'Department', 'Salary']]
```

```python
df.sort_values(['Salary'], ascending=False)
```

```python
avg_salary = df.groupby('Gender')['Salary'].mean()
diff = avg_salary['M'] - avg_salary['F']
print("男女平均薪资差异为：", diff)
```
