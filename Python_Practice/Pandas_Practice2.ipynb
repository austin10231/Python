```python
import pandas as pd

data = {
    'Date': ['2025-01-01', '2025-01-02', '2025-01-03', '2025-01-04', '2025-01-05',
             '2025-01-06', '2025-01-07', '2025-01-08', '2025-01-09', '2025-01-10'],
    'Store': ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C', 'C'],
    'Product': ['Coffee', 'Tea', 'Juice', 'Coffee', 'Tea', 'Juice', 'Coffee', 'Tea', 'Juice', 'Coffee'],
    'Units_Sold': [30, 45, 25, 40, 50, None, 20, 60, 30, None],
    'Unit_Price': [5.0, 4.0, 6.0, 5.5, 4.5, 6.0, 4.5, 4.0, 6.5, 5.0],
    'Customer_Rating': [4.2, 4.5, 4.0, 4.6, 4.8, 4.2, 3.9, None, 4.3, 4.0]
}

df = pd.DataFrame(data)
```

```python
df.drop(columns='Date',inplace=True)
```

```python
df['Total_Revenue'] = df['Units_Sold'] * df['Unit_Price']
```

```python
df.groupby('Store')['Units_Sold'].nunique()
```

```python
df.dropna(subset='Customer_Rating', inplace=True)
```

```python
df['Units_Sold'] = df['Units_Sold'].fillna(df['Units_Sold'].mean())
```

```python
df['Total_Revenue'] = df['Total_Revenue'].fillna(df['Total_Revenue'].mean())
```

```python
def Performance(row):
    if row['Customer_Rating'] >= 4.5 and row['Units_Sold'] >= 40:
        return "Excellent"
    elif row['Customer_Rating'] >= 4.0:
        return "Good"
    else:
        return "Needs Improvement"

df['Performance'] = df.apply(Performance, axis=1)
```

```python
df
```

```python
import matplotlib.pyplot as plt

store_revenue = df.groupby('Store')['Total_Revenue'].sum()

store_revenue.plot(kind='bar', color='skyblue', figsize=(6,4))
plt.title('Total Revenue by Store')
plt.xlabel('Store')
plt.ylabel('Total Revenue($)')
plt.show()
```
