```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
crash_df = sns.load_dataset('car_crashes')
crash_df.head()
```

#### Distribution Plot

```python
sns.distplot(crash_df['not_distracted'], kde=False, bins=25)
```

#### Joint Plot

```python
sns.jointplot(x='speeding', y='alcohol', data=crash_df, kind='kde')
```

#### KDE plot

```python
sns.kdeplot(crash_df['alcohol'])
```

#### Pair Plots

```python
tips_df = sns.load_dataset('tips')

sns.pairplot(tips_df, hue='sex', palette='Blues')
```

#### Rug plot

```python
sns.rugplot(tips_df['tip'])
```

#### Styling

```python
sns.set_style('ticks')

plt.figure(figsize=(8,4))

sns.set_context('paper', font_scale=1.5)

sns.jointplot(x='speeding', y='alcohol', data=crash_df, kind='reg')

sns.despine(left=True, bottom=True)
```

#### Bar Plot

```python
sns.barplot(x='sex', y='total_bill', hue='sex', data=tips_df, estimator = np.cov)
```

#### Count plot

```python
sns.countplot(x='sex', data=tips_df)
```

#### Box plot

```python
sns.boxplot(x='day', y='total_bill', data=tips_df, hue='sex')

plt.legend(loc=0)
```

#### Violin Plot

```python
sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex', split=True)
```

#### Strip Plot

```python
plt.figure(figsize=(8,5))

sns.stripplot(x='day', y='total_bill', data=tips_df, jitter=True, hue='sex', dodge=True)
```

#### Swarm Plot

```python
sns.violinplot(x='day', y='total_bill', data=tips_df)
sns.swarmplot(x='day', y='total_bill', data=tips_df)
```

#### Palettes

```python
plt.figure(figsize=(8,5))
sns.set_style('white')
sns.set_context('talk')
sns.stripplot(x='day', y='total_bill', data=tips_df, hue='sex', palette='afmhot')
plt.legend(loc=3)
```

#### Matrix Plots

#### Heatmaps

```python
plt.figure(figsize=(8,6))
sns.set_context('paper', font_scale=1.4)

crash_mx = crash_df.select_dtypes(include='number').corr()
sns.heatmap(crash_mx, annot=True, cmap='Blues')
```

```python
flights = sns.load_dataset('flights')
flights = flights.pivot_table(index='month', columns='year', values='passengers')
sns.heatmap(flights, cmap='Reds', linecolor='white', linewidth=1)
```

#### Cluster Map

```python
# iris = sns.load_dataset('iris')

# species = iris.pop('species')
# sns.clustermap(iris)

sns.clustermap(flights, cmap='Blues', standard_scale=1)
```

#### PairGrid

```python
iris = sns.load_dataset('iris')
iris_g = sns.PairGrid(iris, hue="species")
iris_g.map_diag(plt.hist)
# iris_g.map_offdiag(plt.scatter)
iris_g.map_upper(plt.scatter)
iris_g.map_lower(sns.kdeplot)
iris_g = sns.PairGrid(iris, hue='species',
                     x_vars=["sepal_length", "sepal_width"],
                     y_vars=["sepal_length", "sepal_width"])
iris_g.map(plt.scatter)
iris_g.add_legend()
```
